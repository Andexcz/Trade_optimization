### Profit Function
$$P(x) = \alpha x - \beta x^2 - cx$$

### Arguments
* **$\alpha$** : Expected return per share
* **$\beta$** : Market impact coefficient (e.g., price slippage due to low liquidity)
* **$c$** : Transaction cost per share
* **$x$** : Number of shares to trade

```python
import numpy as np

def profit_analysis(alpha, beta, c):
    # Calculate the optimal number of shares to trade
    x_optimal = (alpha - c) / (2 * beta)
    print(f"Optimal number of shares to trade: {x_optimal}")
    # Calculate the maximum profit at the optimal number of shares
    max_profit = alpha * x_optimal - beta * x_optimal**2 - c * x_optimal
    print(f"Maximum profit at optimal trade: {max_profit}")
    return x_optimal, max_profit
# Example usage
profit_analysis(0.05,0.0001,0.01)

# The second derivative of the profit function P''(x) = -2β
# This confirms that the profit function is concave down, indicating a maximum at the optimal point.
def second_derivative_test(beta):
    second_derivative = -2 * beta
    if second_derivative < 0:
        print("The profit function is concave down at the optimal point, confirming a maximum.")
    else:
        print("The profit function is not concave down at the optimal point.")
    return second_derivative
```

### Second Derivative Check
The second derivative of the profit function is given by:

$$P''(x) = -2\beta$$

Since we assume $\beta > 0$, it follows that $P''(x) < 0$. This confirms that the profit function is **concave down**, ensuring that the critical point is indeed a **global maximum**.

```python
def second_derivative_test(beta):
    second_derivative = -2 * beta
    if second_derivative < 0:
        print("The profit function is concave down at the optimal point, confirming a maximum.")
    else:
        print("The profit function is not concave down at the optimal point.")
    return second_derivative
# Example usage
second_derivative_test(0.0001)
```

### Risk-Penalized Return Function


$$R(\ell) = \ell\mu - \frac{1}{2}\ell^2\sigma^2$$

**Arguments:**
* $\mu$ : Expected return
* $\sigma$ : Volatility 
* $\ell$ : Leverage multiplier


### Finding Optimal Leverage

Taking the derivative and setting it to zero to find the maximum:

$$R'(\ell) = \mu - \ell\sigma^2 = 0$$

Solving for the optimal leverage $\ell^*$:

$$\ell^* = \frac{\mu}{\sigma^2}$$

The simplest form of **Kelly Criterion**

```python
def kelly_analysis(mu, sigma):
    Optimal_leverage= mu / (sigma ** 2)
    RA_return=mu-(Optimal_leverage*(sigma**2))/2
    print(f"Optimal leverage: {Optimal_leverage}")
    print(f"Risk-adjusted return at optimal leverage: {RA_return}")
# Example usage
kelly_analysis(0.08,0.2)
```

### Expected Value Model

$$E(s) = p(s) \cdot G - (1 - p(s)) \cdot s$$

**Arguments:**
* $p(s)$ : Probability of winning (reaching profit target before stop)
* $G$ : Gain if you win (fixed target)
* $s$ : Loss if stopped out


```python
def Expectations(p,G,s):
    E=p*G-(1-p)*s
    print(f"Expected value of the trade: {E}")
    return E
# Example usage
E = Expectations(0.37,45,30)
if E > 0:
    print("The trade has a positive expected value.")
elif E < 0:
    print("The trade has a negative expected value.")
elif E == 0:
    print("The trade has a zero expected value.")
```
